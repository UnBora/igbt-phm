# Phase 3 — Exploratory Data Analysis (EDA)

## 1. EDA មានន័យថាអ្វី?

**EDA = Exploratory Data Analysis**

មានន័យថា៖

> យើងប្រើ **graphs, distributions, boxplots, correlations និង statistics** ដើម្បីស្វែងយល់ថា data មានលក្ខណៈយ៉ាងដូចម្តេច មុនពេលយើងធ្វើ `Data Cleaning`, `Feature Extraction`, `Health Indicator` ឬ `RUL Prediction`។

Flow របស់យើងគឺ៖

```text
Phase 2
MAT files
    ↓
DataFrame
    ↓
Phase 3 — EDA
    ↓
Visualize
    ↓
Observe patterns
    ↓
Ask PHM questions
    ↓
Later decide what is useful
```

# 3. ចំណុចសំខាន់បំផុត — EDA ≠ Degradation Verdict

នេះគឺជាគំនិតដែលអ្នកគួរចងចាំបំផុត។

បើយើងឃើញ graph មួយមាន trend៖

```text
Signal
  │
  │       /
  │      /
  │    /
  │  /
  │_/
  └──────── Time
```

យើង **មិនអាចនិយាយភ្លាមៗថា**៖

> "This is IGBT degradation."

បានទេ។

ហេតុអ្វី?

ព្រោះ dataset នេះមិនមែនជា experiment ដែលមាន **constant operating condition** ទេ។

វាជា៖

> **temperature-controlled stress experiment**

មានន័យថា `Temperature`, `Stress`, ឬ operating conditions អាចត្រូវបានផ្លាស់ប្តូរតាម experiment protocol។

ដូច្នេះ៖

```text
Observed Trend
      ↓
Could be degradation
      ↓
BUT could also be
      ├── Temperature change
      ├── Stress change
      ├── Operating condition
      ├── Measurement drift
      ├── Missing transient data
      └── Scaling problem
```

---

# 4. ហេតុអ្វី Temperature សំខាន់?

សម្រាប់ IGBT aging experiment យើងកំពុងសិក្សា behavior ក្រោម **controlled stress conditions**។

ឧទាហរណ៍ ប្រសិនបើ `Temperature` កើន ហើយ `Collector Current` ក៏កើន៖

```text
Temperature ↑
      ↓
Collector Current ↑
```

យើងមិនអាចនិយាយថា៖

> Current increased because the IGBT degraded.

ប្រហែលជា៖

> Current increased because the experimental temperature/stress condition changed.

ដូច្នេះ EDA គឺជាការសួរ **"What else could explain this pattern?"**

# 6. Figure 2 — Histograms

# 8. Figure 4 — Pearson Correlation Matrix

ទីបួន យើងមើល **Pearson correlation** រវាង measured signals។

គំនិតសាមញ្ញ៖

```text
Signal A ↑
     │
     │
     └──── Signal B ↑
```

បើ signals ផ្លាស់ប្តូរទិសដៅដូចគ្នា → **positive correlation**

បើមួយកើន ហើយមួយទៀតថយ → **negative correlation**

បើមិនមាន linear relationship ច្បាស់ → correlation ជិត 0។

### ប៉ុន្តែចងចាំ៖

> **Correlation ≠ Causation**

មានន័យថា បើ `Temperature` និង `Current` មាន correlation ខ្ពស់ វាមិនមានន័យថា Temperature គឺជា **cause** របស់ Current ដោយផ្ទាល់ទេ។

វាអាចមាន **third factor** ដែលធ្វើឱ្យ signals ទាំងពីរផ្លាស់ប្តូរដូចគ្នា។

## ⭐ ចំណុចសំខាន់បំផុតសម្រាប់អ្នក

អ្នកអាចចងចាំ Phase 3 ជា **4 words**:

> **TIME → DISTRIBUTION → FILE → CORRELATION**

| EDA             | យើងចង់ដឹង                                         |
| --------------- | ------------------------------------------------- |
| **Time plots**  | Signal ផ្លាស់ប្តូរតាមពេលវេលាយ៉ាងដូចម្តេច?         |
| **Histograms**  | Signal values មាន distribution យ៉ាងដូចម្តេច?      |
| **Boxplots**    | Files/segments មានភាពខុសគ្នាយ៉ាងដូចម្តេច?         |
| **Correlation** | Signals មាន linear relationship គ្នាយ៉ាងដូចម្តេច? |

ហើយរាល់ graph ត្រូវសួរខ្លួនឯង៖

> **“What does this plot tell me about degradation, and what else could explain the pattern?”**
