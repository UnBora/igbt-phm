# 2. តើ Feature ល្អសម្រាប់ PHM គួរមានអ្វីខ្លះ?

ឧទាហរណ៍៖

> ប្រសិនបើ degradation ធ្វើឱ្យ switching voltage peak ផ្លាស់ប្តូរ → voltage peak អាចជាបេក្ខជន feature ដែលគួរសិក្សា។

### ② Measurable Reliably

### ③ Comparable

យើងអាចសិក្សាបានថា RMS មានការផ្លាស់ប្តូរឬអត់។

# 6. `dt` គឺជាអ្វី?

រាល់ transient capture មាន `dt`។

`dt` = **sampling time interval**

មានន័យថា ចន្លោះពេលរវាង samples។

ឧទាហរណ៍៖

```text
Sample 1 → t = 0
Sample 2 → t = dt
Sample 3 → t = 2dt
Sample 4 → t = 3dt
```

បើ៖

```text
dt = 1 μs
```

មានន័យថា samples ត្រូវបានកត់ត្រារៀងរាល់ 1 microsecond។

# 9. Mean

**Mean** = average value។

ឧទាហរណ៍៖

```text
Values = 2, 3, 4

Mean = 3
```

វាប្រាប់យើងអំពី **average level / offset** របស់ signal។

---

# 10. Standard Deviation

**Standard Deviation (Std)** បង្ហាញថា values រាលដាលពី mean ប៉ុន្មាន។

ឧទាហរណ៍៖

```text
Signal A:
10, 10, 10, 10
→ low variation

Signal B:
5, 10, 15, 20
→ high variation
```

ដូច្នេះ Std ជួយពិពណ៌នា **spread/variation** របស់ waveform។

---

# 11. RMS

**RMS = Root Mean Square**

វាជា measurement ដែលពេញនិយមសម្រាប់សង្ខេប magnitude របស់ signal។

Conceptually៖

```text
Raw signal
    ↓
Square
    ↓
Average
    ↓
Square root
    ↓
RMS
```

វាអាចមានប្រយោជន៍ក្នុងការពិពណ៌នា signal magnitude។

---

# 12. Minimum និង Maximum

យើងរក៖

```text
Minimum
Maximum
```

ឧទាហរណ៍៖

```text
Signal:
-2, 1, 5, 8, 3

Min = -2
Max = 8
```

វាជួយយើងស្គាល់ **extrema** របស់ waveform។

---

# 13. Peak-to-Peak

**Peak-to-Peak (P-P)** គឺ៖

> Maximum − Minimum

ឧទាហរណ៍៖

```text
Max = 8
Min = -2

Peak-to-Peak = 10
```

វាពិពណ៌នា **total amplitude range** របស់ capture។

---

# 14. Skewness

**Skewness** ពិពណ៌នាអំពីភាពមិនស្មើគ្នារបស់ distribution។

Conceptually៖

```text
Normal-ish
      /\
     /  \
____/    \____

Skewed
       /\
      /  \
_____/    \______
```

វាផ្តល់ information អំពី **shape** របស់ signal distribution។

---

# 15. Excess Kurtosis

**Excess Kurtosis** ជួយពិពណ៌នា **tail/peakedness** របស់ distribution។

វាអាចជួយយើងសង្កេតថា signal មាន extreme values ឬ behavior ខុសពី distribution ធម្មតាយ៉ាងដូចម្តេច។

---

# 16. Crest Factor

**Crest Factor** ជាទូទៅពិពណ៌នាទំនាក់ទំនងរវាង peak និង RMS៖

> Peak magnitude relative to RMS

Conceptually៖

```text
Large peak
    ↓
Crest Factor ↑
```

វាជា descriptor មួយសម្រាប់ signal **shape/amplitude characteristics**។

---

# 17. ហេតុអ្វីជ្រើស Features ទាំងនេះ?

Features ទាំងនេះអាចសង្ខេប 4 ប្រភេទសំខាន់ៗ៖

| Feature group      | Examples                         | អ្វីដែលវាពិពណ៌នា            |
| ------------------ | -------------------------------- | --------------------------- |
| **Offset / Level** | Mean                             | Average level               |
| **Spread**         | Std, RMS                         | Variation / magnitude       |
| **Amplitude**      | Min, Max, Peak-to-Peak           | Signal range                |
| **Shape**          | Skewness, Kurtosis, Crest Factor | Waveform/distribution shape |

ដូច្នេះយើងមិនគ្រាន់តែមាន raw waveform ទេ ប៉ុន្តែមាន **compact numerical description**។

# 19. Peak-to-Peak ≠ Turn-Off Peak

នេះជាចំណុចដែលគួរយល់ច្បាស់។

បើយើងគណនា៖

```text
Peak-to-Peak = Max - Min
```

វាជា feature របស់ **whole capture**។

វា **មិនដូចជា** ការដកស្រង់ **specific turn-off peak** ដែលបានកំណត់ទីតាំងត្រឹមត្រូវទេ។

ឧទាហរណ៍៖

```text
Whole capture
────────────────────────
       /\         /\
      /  \_______/  \
_____/              \____
          ↑
      turn-off event
```

`Whole-capture peak-to-peak` អាចរួមបញ្ចូល behavior ផ្សេងៗជាច្រើន។
