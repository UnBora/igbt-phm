# 1. Health Indicator (HI) គឺជាអ្វី?

**Health Indicator (HI)** គឺជា quantity មួយដែលយើងវាស់ ឬគណនាចេញពី data ដើម្បី **តាមដាន health state របស់ system តាមពេលវេលា**។

សាមញ្ញៗ៖

```text
Healthy
   ↓
Aging
   ↓
Degradation
   ↓
More degraded
   ↓
Failure / EOL
```

HI គួរតែមានសមត្ថភាពបង្ហាញការផ្លាស់ប្តូរនេះ។

ឧទាហរណ៍ conceptually៖

```text
Health
  │
  │\
  │ \
  │  \
  │   \
  │    \____
  └──────────── Time
      Aging
```

បើ HI មាន relationship ដែលសមហេតុផលជាមួយ degradation យើងអាចប្រើវាសម្រាប់ PHM បាន។

---

# 2. Feature និង Health Indicator មិនដូចគ្នាទេ

នេះជាចំណុចសំខាន់បំផុតពី Phase 5 → Phase 6។

### Feature

គឺ numerical description របស់ data។

ឧទាហរណ៍៖

```text
RMS
Mean
Peak-to-Peak
Standard Deviation
Crest Factor
```

### Health Indicator

គឺ feature ឬ derived quantity ដែលមាន evidence គ្រប់គ្រាន់ថា៖

> **វាអាចតាមដាន health/degradation របស់ system បាន។**

ដូច្នេះ៖

```text
Raw waveform
      ↓
Feature Engineering
      ↓
Feature
      ↓
Validation
      ↓
Candidate HI
      ↓
Physical evidence
      ↓
Credible HI
```

មិនមែន feature ទាំងអស់សុទ្ធតែជា HI ទេ។

# 3. HI ល្អគួរមានលក្ខណៈអ្វីខ្លះ?

Text នេះផ្តល់ criteria សំខាន់ៗជាច្រើន។

## ① Plausible link to degradation mechanism

HI គួរតែមានទំនាក់ទំនងដែលសមហេតុផលជាមួយ **physical degradation mechanism**។

ឧទាហរណ៍៖

```text
IGBT aging
    ↓
Physical property changes
    ↓
Switching behavior changes
    ↓
Voltage transient changes
    ↓
Candidate HI
```

យើងត្រូវអាចពន្យល់បានថា **ហេតុអ្វី** physical degradation អាចធ្វើឱ្យ measurement នោះផ្លាស់ប្តូរ។

# 4. Consistent response

នៅពេល damage កើនឡើង HI គួរមាន response ដែល **consistent**។

ឧទាហរណ៍៖

```text
Aging ↑
   ↓
HI ↑
```

ឬ៖

```text
Aging ↑
   ↓
HI ↓
```

ទាំងពីរអាចត្រឹមត្រូវ។

សំខាន់គឺ៖

> **យើងត្រូវដឹងថា direction នោះមានន័យអ្វី និងមាន physical justification។**

---

# 5. Monotonicity

**Monotonicity** មានន័យថា HI មាន tendency ផ្លាស់ប្តូរទៅក្នុងទិសដៅមួយនៅពេល degradation កើនឡើង។

ឧទាហរណ៍៖

```text
HI
│
│ •
│  •
│   •
│    •
│     •
└────────── Time
```

នេះមាន **decreasing trend**។

ឬ៖

```text
HI
│      •
│     •
│   •
│  •
│ •
└────────── Time
```

នេះមាន **increasing trend**។

ប៉ុន្តែ real-world data អាចមាន noise៖

```text
HI
│     •
│   •   •
│  • •
│ •    •
│• 
└────────── Time
```

ដូច្នេះ monotonicity មិនមានន័យថា value ត្រូវកើន/ថយរាល់ observation ទេ។ យើងចង់ឃើញ **overall consistent direction**។

---

# 6. Sensitivity

**Sensitivity** មានន័យថា HI អាច detect ការផ្លាស់ប្តូរតូចៗក្នុង health/degradation បានឬទេ។

ឧទាហរណ៍៖

### Feature A

```text
Healthy → Aging

10.00
10.01
10.00
10.02
10.01
```

វាអាចមិនសូវ sensitive។

### Feature B

```text
Healthy → Aging

10
11
13
15
18
```

វាបង្ហាញការផ្លាស់ប្តូរច្បាស់ជាង។

ប៉ុន្តែ **ការផ្លាស់ប្តូរច្រើន ≠ ល្អជានិច្ច**។ វាអាចជា noise ឬ operating-condition effect ផងដែរ។

# 7. Manageable Noise

```text
Bad candidate
│ •  •
│    •
│ •    •
│      •
│  •
└──────── Time
```

នេះសំខាន់ខ្លាំងសម្រាប់ NASA IGBT dataset ព្រោះមាន៖

# 9. Statistical Trend ≠ Good HI

នេះជាគំនិតសំខាន់បំផុតមួយក្នុង Phase 6។

Feature អាចមាន beautiful trend៖

```text
HI
│       •
│     •
│   •
│ •
│•
└──────── Time
```

ប៉ុន្តែវាអាចជា **poor HI**។

ហេតុអ្វី?

ព្រោះ trend អាចកើតពី៖

```text
Trend
  │
  ├── Device degradation
  ├── Load change
  ├── Temperature set-point
  ├── Instrument drift
  ├── Acquisition setting
  └── File boundary
```

ដូច្នេះ៖

> **Statistical trend alone is not enough.**
