# 3. MATLAB `.mat` File គឺជាអ្វី?

`.mat` គឺជា file format របស់ **MATLAB**។

វាមិនមានន័យថា file នោះជា table ដូចជា Excel/CSV ទេ។

```text
CSV
 ├── Time
 ├── Voltage
 ├── Current
 └── Temperature
```

អាចមើលទៅដូចជា table។

ប៉ុន្តែ `.mat` file អាចមាន **nested structures**៖

```text
MAT file
   │
   └── measurement
          │
          └── steadyState
                 │
                 ├── record 1
                 ├── record 2
                 ├── record 3
                 └── ...
```

# 14. Data Leakage

Text នេះនិយាយថា source identity មានសារៈសំខាន់ដើម្បីជៀសវាង **data leakage**។

`Data leakage` មានន័យថា information ពី data ដែលមិនគួរចូលក្នុងការវាយតម្លៃ ឬ training ត្រូវបានលេចចូលក្នុង model process។

ឧទាហរណ៍៖

```text
Training Data
     ↓
Model
     ↑
Test information accidentally included
```

នេះអាចធ្វើឱ្យ model មើលទៅមាន performance ល្អ ប៉ុន្តែ performance នោះមិនត្រឹមត្រូវ។

ដូច្នេះការរក្សា `source_file` និង `device` identity ជួយឱ្យយើងយល់ពីប្រភព និង test segments មុនពេល modeling។
